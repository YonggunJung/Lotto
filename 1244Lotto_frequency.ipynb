```python
import pandas as pd
import numpy as np
from collections import Counter
from google.colab import files
```

```python
# CSV 업로드
# uploaded = files.upload()
import io
df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/로또/data/jlotto.csv', header=None)
df
```

```python
# 번호 데이터 추출 및 이상치(45보다 큰 값) 보정
# 마지막 행의 338을 38로 수정하거나 이상 데이터를 정제합니다.
lotto_data = df.values[:]

# 각 행에서 1~45 범위를 벗어나는 값(예: 338)이 있으면 적절한 값(예: 38)으로 변경
refined_lotto_data = []
for row in lotto_data:
    refined_row = []
    for val in row:
        val = int(val)
        if val > 45:
            # 338 같은 이상치의 경우 일시적으로 38로 보정하거나 필요시 데이터를 수정하세요.
            print(f"이상치 발견 및 보정: {val} -> {val % 100 if val % 100 <= 45 else 38}")
            val = val % 100 if val % 100 <= 45 else 38
        refined_row.append(val)
    refined_lotto_data.append(refined_row)

lotto_data = refined_lotto_data
lotto_data[-5:]  # 끝부분 데이터 정상 보정 여부 확인
```

```python
# 빈도수 행렬 초기화 (45x45)
# freq_matrix[i][j]는 숫자 i+1이 포함된 행 다음 줄에서 숫자 j+1이 몇 번 나왔는지
freq_matrix = np.zeros((45, 45), dtype=int)
freq_matrix
```

```python
# 빈도수 행렬 초기화 후 다시 계산
freq_matrix = np.zeros((45, 45), dtype=int)

for i in range(len(lotto_data) - 1):  # 마지막 줄은 다음 줄이 없어서 제외
    current_row = lotto_data[i]
    next_row = lotto_data[i + 1]

    for num in current_row:
        for next_num in next_row:
            freq_matrix[num - 1][next_num - 1] += 1
freq_matrix
```
